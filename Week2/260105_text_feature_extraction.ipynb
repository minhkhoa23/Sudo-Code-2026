```python
import pandas as pd
import sklearn
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
```

```python
df = pd.read_json('/kaggle/input/datasets/haitranquangofficial/vietnamese-online-news-dataset/news_dataset.json')
```

```python
df.head()
```

## Bag of word

```python
bow_vectorizer = CountVectorizer()

bow = bow_vectorizer.fit_transform(df['content'])

bow.shape
```

```python
# bow.toarray()[:30, :30]
```

```python
bow_feature = bow_vectorizer.get_feature_names_out()

bow_feature[:100]
```

## TF - IDF

```python
tfidf_vectorizer = TfidfVectorizer()

tfidf = tfidf_vectorizer.fit_transform(df['content'])

tfidf.shape
```

```python
# tfidf.toarray()[:30, :30]
```

```python
tfidf_feature = tfidf_vectorizer.get_feature_names_out()

tfidf_feature[:30]
```

## Uni-gram

```python
uni_vectorizer = CountVectorizer(ngram_range=(1, 1))

unigram = uni_vectorizer.fit_transform(df['content'])

unigram.shape
```

```python
# unigram.toarray()[:30, :30]
```

```python
unigram_feature = uni_vectorizer.get_feature_names_out()

unigram_feature[:30]
```

## Bi-gram

```python
bi_vectorizer = CountVectorizer(ngram_range=(2, 2))

bigram = bi_vectorizer.fit_transform(df['content'])

bigram.shape
```

```python
# bigram.toarray()[:30, :30]
```

```python
bigram_feature = bi_vectorizer.get_feature_names_out()

bigram_feature[:30]
```
